Here are a few random computations I ran with SageMath while thinking about my Linear Algebra class.

```python
set_random_seed(1)
M = random_matrix(ZZ, 5,5)
M[0] = M[1] + M[3]
M = M.transpose()
[list(r) for r in M.rows()]
```

```python
latex(M)
```

```python
set_random_seed(2)
M = random_matrix(ZZ, 5,5, x=-8, y=9)
M[2] = M[0] - 2*M[1]
M[4] = M[0]+M[1]+M[3]
M = M.transpose()
M.transpose().rref()

print(latex(M))
[list(r) for r in M.rows()]
```
