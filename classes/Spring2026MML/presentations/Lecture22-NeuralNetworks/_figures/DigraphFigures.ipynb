## An empty NN

```python
g = Digraph(engine='neato')
g.attr(size='8,5!')
g.attr(ratio='0.7')
nodes0 = []
nodes1 = []
nodes2 = []
output_nodes = []

# g.node('label_input0', label='Input layer', shape='plaintext', fontsize='20', pos='0,4.6!')
for i in range(0,4):
    node0 = g.node(
        f'x0,{i}', pos=f'0,{i}!', shape="circle", label="", fontsize='18'
    )
    nodes0.append(f'x0,{i}')

# g.node('label_input1', label='Hidden layer 1', shape='plaintext', fontsize='20', pos='2,4.6!')
for i in range(0,6):
    g.node(f'x1,{i}', pos=f'2,{i-1}!', shape="circle", label="")
    nodes1.append(f'x1,{i}')
# g.node('label_input2', label='Hidden layer 2', shape='plaintext', fontsize='20', pos='4,4.6!')
for i in range(0,5):
    g.node(f'x2,{i}', pos=f'4,{i-0.5}!', shape="circle", label="")
    nodes2.append(f'x2,{i}')
# g.node('label_input3', label='Output', shape='plaintext', fontsize='20', pos='6,4.6!')
for i in range(1,3):
    g.node(f'y,{i}', pos = f'6,{i}!', shape="circle", label="")
    output_nodes.append(f'y,{i}')

for i,node0 in enumerate(nodes0):
    for j,node1 in enumerate(nodes1[1:]):
        g.edge(node0,node1, color='#000000aa')
for i,node1 in enumerate(nodes1):
    for j,node2 in enumerate(nodes2[1:]):
        g.edge(node1,node2, color='#000000aa')
for i,node2 in enumerate(nodes2):
    for j,output_node in enumerate(output_nodes):
        g.edge(node2,output_node, color='#000000aa')
g.render("empty_neural_network", format='png')
g
```

## A labeled NN

```python
g = Digraph(engine='neato')
g.attr(size='8,5!')
g.attr(ratio='0.7')
nodes0 = []
nodes1 = []
nodes2 = []
output_nodes = []

g.node('label_input0', label='Input layer', shape='plaintext', fontsize='20', pos='0,4.6!')
for i in range(0,4):
    node0 = g.node(
        f'x0,{i}', pos=f'0,{i}!', shape="circle", fontsize='18', 
        label="1" if i == 0 else f"<<i>x</i><sub>{i}</sub>>"
    )
    nodes0.append(f'x0,{i}')

g.node('label_input1', label='Hidden layer 1', shape='plaintext', fontsize='20', pos='2,4.6!')
for i in range(0,6):
    g.node(f'x1,{i}', pos=f'2,{i-1}!', shape="circle", fontsize='18', label="1" if i == 0 else f"<<i>h</i><sub>1,{i}</sub>>")
    nodes1.append(f'x1,{i}')
g.node('label_input2', label='Hidden layer 2', shape='plaintext', fontsize='20', pos='4,4.6!')
for i in range(0,5):
    g.node(f'x2,{i}', pos=f'4,{i-0.5}!', shape="circle", fontsize='18', label="1" if i == 0 else f"<<i>h</i><sub>2,{i}</sub>>")
    nodes2.append(f'x2,{i}')
g.node('label_input3', label='Output', shape='plaintext', fontsize='20', pos='6,4.6!')
for i in range(1,3):
    g.node(f'y,{i}', pos = f'6,{i}!', shape="circle", fontsize='18', label=f"<<i>y</i><sub>{i}</sub>>")
    output_nodes.append(f'y,{i}')

for i,node0 in enumerate(nodes0):
    for j,node1 in enumerate(nodes1[1:]):
        if (i==0 and j==0) or (i==3 and j==4):
            label = f"<<i>w</i><sub>0,{i},{j+1}</sub>>"
        else:
            label = ""
        g.edge(node0,node1, color='#000000aa', label=label)
for i,node1 in enumerate(nodes1):
    for j,node2 in enumerate(nodes2[1:]):
        if (i==5 and j==3):
            label = f"<<i>w</i><sub>1,{i},{j+1}</sub>>"
        else:
            label = ""
        g.edge(node1,node2, color='#000000aa', label=label)
for i,node2 in enumerate(nodes2):
    for j,output_node in enumerate(output_nodes):
        if (i==4 and j==1):
            label = f"<<i>w</i><sub>2,{i},{j+1}</sub>>"
        else:
            label = ""
        g.edge(node2,output_node, color='#000000aa', label=label)
g.render("labeled_neural_network", format='pdf')
g
```

## Conflating notation

```python
g = Digraph(engine='neato')
g.attr(size='8,5!')
g.attr(ratio='0.7')
nodes0 = []
nodes1 = []
nodes2 = []
output_nodes = []

g.node('label_input0', label='Layer 0', shape='plaintext', fontsize='20', pos='0,4.6!')
for i in range(0,4):
    node0 = g.node(
        f'x0,{i}', pos=f'0,{i}!', shape="circle", fontsize='18', width="0.9", fixedsize='true',
        label=f"<<i>x</i><sub>0,{i}</sub>=1>" if i == 0 else f"<<i>x</i><sub>0,{i}</sub>>"
    )
    nodes0.append(f'x0,{i}')

g.node('label_input1', label='Layer 1', shape='plaintext', fontsize='20', pos='2,4.6!')
for i in range(0,6):
    g.node(
        f'h1,{i}', pos=f'2,{i-1}!', shape="circle", fontsize='18', width="0.9", fixedsize='true',
        label=f"<<i>x</i><sub>1,{i}</sub>=1>" if i == 0 else f"<<i>x</i><sub>1,{i}</sub>>"
    )
    nodes1.append(f'h1,{i}')
    
g.node('label_input2', label='Layer 2', shape='plaintext', fontsize='20', pos='4,4.6!')
for i in range(0,5):
    g.node(
        f'h2,{i}', pos=f'4,{i-0.5}!', shape="circle", fontsize='18', width="0.9", fixedsize='true', 
        label=f"<<i>x</i><sub>2,{i}</sub>=1>" if i == 0 else f"<<i>x</i><sub>2,{i}</sub>>"
    )
    nodes2.append(f'h2,{i}')

g.node('label_input3', label='Layer 3', shape='plaintext', fontsize='20', pos='6,4.6!')
for i in range(1,3):
    g.node(
        f'x,{i}', pos = f'6,{i}!', shape="circle", fontsize='18', 
        label=f"<<i>x</i><sub>3,{i}</sub>>"
    )
    output_nodes.append(f'x,{i}')

for i,node0 in enumerate(nodes0):
    for j,node1 in enumerate(nodes1[1:]):
        if (i==0 and j==0) or (i==3 and j==4):
            label = f"<<i>w</i><sub>0,{i},{j+1}</sub>>"
        else:
            label = ""
        g.edge(node0,node1, color='#000000aa', label=label)
for i,node1 in enumerate(nodes1):
    for j,node2 in enumerate(nodes2[1:]):
        if (i==5 and j==3):
            label = f"<<i>w</i><sub>1,{i},{j+1}</sub>>"
        else:
            label = ""
        g.edge(node1,node2, color='#000000aa', label=label)
for i,node2 in enumerate(nodes2):
    for j,output_node in enumerate(output_nodes):
        if (i==4 and j==1):
            label = f"<<i>w</i><sub>2,{i},{j+1}</sub>>"
        else:
            label = ""
        g.edge(node2,output_node, color='#000000aa', label=label)
g.render("conflated_neural_network", format='png', view=True)
g
```

```python
import numpy as np

sigmoid = lambda x: 1/(1+np.exp(-x))
relu = lambda x: np.maximum(0, x)

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

# input -> hidden
# h1 = ReLU(x1 - x2)
# h2 = ReLU(x2 - x1)
W0 = np.array([
    [ 0,  0],   # bias
    [ 1, -1],   # x1
    [-1,  1]    # x2
])

# hidden -> output
# z = -0.5 + h1 + h2
W1 = np.array([
    [-0.5],
    [ 1.0],
    [ 1.0]
])

X0 = np.column_stack([np.ones(len(X)), X])
X1 = relu(X0 @ W0)
X1 = np.column_stack([np.ones(len(X1)), X1])
X2 = sigmoid(X1 @ W1)

print(X1)
print(X2.flatten())
print(np.round(X2).flatten())
```

```python


import matplotlib.pyplot as plt
def f(x,y):
    XInput = np.array([x,y])
    W0 = np.array([
        [1,-1],
        [-1,1]
    ])
    B0 = np.array([0,0])
        
    W1 = np.array([1,1])
    B1 = np.array([-0.5])
    
    sigmoid = lambda x: 1/(1+np.exp(-x))
    a = relu(np.dot(XInput, W0) + B0)
    yOutput = sigmoid(np.dot(a, W1) + B1)
    
    return yOutput[0]

delta = 0.01
xx = np.arange(-0.3,1.31, delta)
yy = np.arange(-0.3, 1.31, delta)
X, Y = np.meshgrid(xx, yy)
Z = [[f(x,y) for x in xx] for y in yy]

plt.figure(figsize=(5.5,5.5))
ax = plt.gca()
CS = ax.contour(X, Y, Z, 2)
ax.clabel(CS, fontsize=10)
ax.set_title('Classification contours before rounding');

points = [(0, 0), (1, 0), (0, 1), (1, 1)]
for i, (x, y) in enumerate(points):
    plt.plot(x, y, 'o', markersize=8, label=f'({x}, {y})', color='blue' if (x==1 and y==0) or (x==0 and y==1) else 'red')
    plt.text(x + 0.05, y + 0.05, f'({x}, {y})', fontsize=12)
plt.xlabel('$x_1$-axis', fontsize=12)
plt.ylabel('$x_2$-axis', fontsize=12);
```
