```python
x,y = var('x,y')
eqs = [x^2-3*x*y==1, y^2-3*x*y==1]
sols = solve(eqs, [x,y])
sols = [tuple([s.rhs() for s in sol]) for sol in sols
    if all(eq.rhs().is_real() for eq in sol)]
sols
```
